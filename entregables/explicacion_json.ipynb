## Explicación del último modelo rechazado del training_history.json

Para conseguir que el sistema generara versiones rechazadas, fue necesario diseñar un conjunto de entrenamiento que redujera claramente la accuracy del nuevo modelo respecto al modelo activo.

El sistema utiliza una regla de activación basada en la comparación de accuracies: el nuevo modelo solo se activa si su rendimiento es mayor o igual que el anterior. Por tanto, para provocar un rechazo era necesario que la nueva accuracy fuera significativamente menor.

En este caso, se construyó un conjunto de 20 muestras en el que se introdujeron contradicciones explícitas en las etiquetas. Concretamente, se incluyeron ejemplos con exactamente las mismas características (mismos valores de sépalo y pétalo) pero con etiquetas distintas. Desde el punto de vista del modelo, esto supone un conflicto imposible de resolver: dos muestras idénticas no pueden pertenecer a dos clases diferentes al mismo tiempo.

Al entrenar con estos datos y realizar la validación (que en el código se hace mediante un train/test split cuando hay 20 o más muestras), el modelo inevitablemente comete errores en el conjunto de validación. Esto reduce la accuracy del nuevo modelo por debajo de la del modelo activo anterior.

Como consecuencia, el mecanismo de gating detecta que: accuracy_new < accuracy_previous y el sistema registra esa versión como "rechazado" en el training_history.json, manteniendo el modelo anterior como activo.
